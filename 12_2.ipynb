```python
import requests
from bs4 import BeautifulSoup
from textblob import TextBlob
```

```python
response = requests.get('https://www.python.org')
```

```python
soup = BeautifulSoup(response.content, 'html5lib')
```

```python
text = soup.get_text(strip=True)
```

```python
blob = TextBlob(text)
```

```python
sentences = blob.sentences

print("Number of sentences:", len(sentences))

for sentence in sentences[:10]:
    print(sentence)
```

```python
words = blob.words

print("Number of words:", len(words))
print(words[:50])
```

```python
noun_phrases = blob.noun_phrases

print("Number of noun phrases:", len(noun_phrases))

for phrase in noun_phrases[:30]:
    print(phrase)
```

```python
### In this exercise, I used TextBlob to process the text collected from the Python website.I tokenized the text into sentences and words, and then extracted noun phrases from the text.
```
